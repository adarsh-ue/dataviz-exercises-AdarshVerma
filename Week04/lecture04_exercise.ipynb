```python
import pandas as pd
import plotly.express as px


# Dataset: Gapminder — GDP, Life Expectancy, Population by Country
# Source: Gapminder Foundation (gapminder.org)

df = px.data.gapminder()
print(f"Loaded: {len(df)} rows")
print(df.head())
```

```python
# explore

print(df.info())
print("Years:", sorted(df['year'].unique()))
print("Continents:", df['continent'].unique())
print(df.describe().round(1))
```

## Task 1 — Scatter: life expectancy change over time

**What to build:** A scatter showing **GDP per capita vs life expectancy** for **two years** (2002 and 2007) to show how both moved — use **colour for year** (just 2 colours), **one continent only**.

Choose any continent except Africa (that was the example). Highlight the change direction.

> 💡 Filter: `df.loc[df['continent'] == 'YOUR_CHOICE']` then filter years


```python
# Task 1
# YOUR CODE HERE

import plotly.graph_objects as go

asia = df[(df["continent"] == "Asia") & (df["year"].isin([2002, 2007]))].copy()
asia["year"] = asia["year"].astype(str)

fig = px.scatter(
    asia, x="gdpPercap", y="lifeExp", color="year",
    hover_name="country", log_x=True,
    color_discrete_map={"2002": "#bdbdbd", "2007": "#1f77b4"},
    labels={"gdpPercap": "GDP per capita (log scale)", "lifeExp": "Life expectancy (years)", "year": "Year"},
    title="Asian countries got richer and lived longer between 2002 and 2007",
)

for c, g in asia.groupby("country"):
    g = g.sort_values("year")
    fig.add_trace(go.Scatter(x=g["gdpPercap"], y=g["lifeExp"], mode="lines",
                             line=dict(color="lightgrey", width=1),
                             showlegend=False, hoverinfo="skip"))

fig.update_traces(marker=dict(size=9), selector=dict(mode="markers"))
fig.update_layout(template="simple_white")
fig.show()
```

## Task 2 — Bubble chart: tell a story

**What to build:** A bubble chart (full 2007 dataset, all countries) where:
- x = GDP per capita (log scale)
- y = life expectancy
- size = population
- colour = ONE continent highlighted (your choice), all others grey
- At least one annotation explaining the highlighted group's story

> This is the grey-and-highlight technique applied to a bubble chart.


```python
# Task 2
# YOUR CODE HERE

import numpy as np

d07 = df[df["year"] == 2007].copy()
d07["group"] = d07["continent"].where(d07["continent"] == "Africa", "Other")

fig = px.scatter(
    d07, x="gdpPercap", y="lifeExp", size="pop", color="group",
    hover_name="country", log_x=True, size_max=50,
    color_discrete_map={"Africa": "#d62728", "Other": "#d9d9d9"},
    category_orders={"group": ["Other", "Africa"]},
    labels={"gdpPercap": "GDP per capita (log scale)", "lifeExp": "Life expectancy (years)", "group": ""},
    title="In 2007, African countries had the lowest life expectancy at almost every income level",
)

sw = d07[d07["country"] == "Swaziland"].iloc[0]
fig.add_annotation(
    x=np.log10(sw["gdpPercap"]), y=sw["lifeExp"],
    text="Swaziland: 40 years, the lowest of all<br>(HIV/AIDS hit southern Africa hardest)",
    showarrow=True, arrowhead=2, ax=120, ay=20, font=dict(color="#d62728"),
)
fig.update_layout(template="simple_white")
fig.show()
```
